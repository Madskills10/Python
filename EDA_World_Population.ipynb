```python
df.corr(numeric_only=True)    
# Compute pairwise correlation between numeric columns in the DataFrame 'df'. 
# 'numeric_only=True' ensures only numeric columns are included in the calculation.
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)    
# Create a heatmap using seaborn to visualize the correlation matrix of numeric columns in 'df'. 
# 'annot=True' displays the correlation values inside each cell of the heatmap.

plt.rcParams['figure.figsize'] = (10, 5)    
# Set the default figure size for plots to 10 inches wide and 5 inches tall.

plt.show()    
# Render and display the heatmap plot
```

```python
df.groupby("Continent").mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)    
# Group the DataFrame 'df' by the 'Continent' column. 
# Compute the mean of all numeric columns for each continent ('numeric_only=True'). 
# Then sort the results by the '2022 Population' column in descending order to show continents with the largest average population first.
```

```python
df[df['Continent'].str.contains('Oceania')]    
# Filter the DataFrame 'df' to return only rows where the 'Continent' column contains the string 'Oceania'
```

```python
df2 = (
    df.groupby("Continent")[df.columns[5:13][::-1]]
      .mean(numeric_only=True)
      .sort_values(by="2022 Population", ascending=False)
)

df2    
# Group the DataFrame 'df' by the 'Continent' column. 
# Select columns 5 through 12 in reverse order (using [::-1]). 
# Compute the mean of these numeric columns for each continent ('numeric_only=True'). 
# Then sort the results by the '2022 Population' column in descending order. 
# Finally, assign the result to 'df2' and display it.
```

```python
df.columns    
# Display the column labels of the DataFrame 'df' as an Index object
```

```python
df3 = df2.transpose()    
# Transpose the DataFrame 'df2', swapping rows and columns so that continents become columns and years become rows.

df3    
# Display the transposed DataFrame 'df3'
```

```python
ax = df3.plot(figsize=(15, 8), marker='o')    
# Plot the transposed DataFrame 'df3' as a line chart with circle markers for each data point. 
# The figure size is set to 15 inches wide and 8 inches tall.

for line in ax.lines:
    for x, y in zip(line.get_xdata(), line.get_ydata()):
        ax.text(x, y, f'{y:,.0f}', ha='center', va='bottom', fontsize=8)    
# Loop through each line in the plot. 
# For every data point (x, y), add a text label showing the value formatted with commas and no decimals. 
# Labels are centered horizontally and placed just above each marker with font size 8.

plt.show()    
# Render and display the annotated line chart
```
